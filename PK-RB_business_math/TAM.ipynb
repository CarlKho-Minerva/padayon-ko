````python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Revenue Model Projections
years = ["Year 1", "Year 2", "Year 3"]
b2c_revenue = [10.3e6, 25.75e6, 51.5e6]  # B2C Revenue in ₱
b2b_revenue = [30e6, 60e6, 90e6]  # B2B Revenue in ₱

# Total Revenue Calculation
total_revenue = [b2c + b2b for b2c, b2b in zip(b2c_revenue, b2b_revenue)]

# Create DataFrame for easier plotting
data = {
    "Year": years,
    "B2C Revenue": b2c_revenue,
    "B2B Revenue": b2b_revenue,
    "Total Revenue": total_revenue,
}
df = pd.DataFrame(data)

# Set the style
sns.set(style="whitegrid")


# Function to create individual bar charts
def create_bar_chart(data, title, ylabel):
    plt.figure(figsize=(8, 5))
    sns.barplot(x="Year", y=data, palette="viridis", data=df)
    plt.title(title, fontsize=16)
    plt.xlabel("Year", fontsize=14)
    plt.ylabel(ylabel, fontsize=14)
    plt.grid(axis="y")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()


# Create individual bar charts for B2C, B2B, and Total Revenue
create_bar_chart(df["B2C Revenue"], "B2C Revenue Projections (₱)", "Revenue (₱)")
create_bar_chart(df["B2B Revenue"], "B2B Revenue Projections (₱)", "Revenue (₱)")
create_bar_chart(df["Total Revenue"], "Total Revenue Projections (₱)", "Revenue (₱)")

# Summary Chart for all Revenues
plt.figure(figsize=(12, 6))
df.plot(
    x="Year",
    kind="bar",
    stacked=False,
    color=["#1f77b4", "#ff7f0e", "#2ca02c"],
    alpha=0.8,
)
plt.title("Revenue Model Summary (₱)", fontsize=16)
plt.xlabel("Year", fontsize=14)
plt.ylabel("Revenue (₱)", fontsize=14)
plt.xticks(rotation=0)
plt.legend(title="Revenue Sources", fontsize=12)
plt.grid(axis="y")

# Display the summary plot
plt.tight_layout()
plt.show()
````

````python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Revenue Model Projections
years = ["Year 1", "Year 2", "Year 3"]
b2c_revenue = [10_300_000, 25_750_000, 51_500_000]  # B2C Revenue in ₱
b2b_revenue = [30_000_000, 60_000_000, 90_000_000]  # B2B Revenue in ₱

# Total Revenue Calculation
total_revenue = [b2c + b2b for b2c, b2b in zip(b2c_revenue, b2b_revenue)]

# Create DataFrame for easier plotting
data = {
    "Year": years,
    "B2C Revenue": b2c_revenue,
    "B2B Revenue": b2b_revenue,
    "Total Revenue": total_revenue,
}
df = pd.DataFrame(data)

# Melt the DataFrame for Seaborn
df_melted = df.melt(
    id_vars="Year",
    value_vars=["B2C Revenue", "B2B Revenue", "Total Revenue"],
    var_name="Revenue Source",
    value_name="₱ Millions",
)

# Set the style and palette
sns.set(style="whitegrid")
sns.set_palette("muted")

# Create a bar plot
plt.figure(figsize=(12, 6))
sns.barplot(x="Year", y="₱ Millions", hue="Revenue Source", data=df_melted, alpha=0.8)

# Chart details
plt.title("Revenue Model Summary (₱)", fontsize=16)
plt.xlabel("Year", fontsize=14)
plt.ylabel("Revenue (₱)", fontsize=14)
plt.xticks(rotation=0)
plt.legend(title="Revenue Sources", fontsize=12)
plt.grid(axis="y")

# Display the summary plot
plt.tight_layout()
plt.show()
````
